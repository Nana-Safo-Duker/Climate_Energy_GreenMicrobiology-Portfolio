# Climate Change Impacts on Renewable Energy Generation

This notebook accompanies the review repository for **"Climate change impacts on renewable energy generation. A review of quantitative projections"**.

It uses manually curated summary values reported in the literature to:

1. Create a compact technology-level dataset.
2. Compute descriptive statistics.
3. Recreate an interpretable summary figure.
4. Connect the quantitative output back to the paper's core conclusions.

> Note: this notebook does not reproduce the paper's full original pipeline because the full gridded climate-model inputs are not bundled in this repository.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

summary_df = pd.DataFrame(
    [
        {"technology": "Utility-scale PV", "absolute_change_ej_per_year": -3.0, "percent_change": -0.4},
        {"technology": "Rooftop PV", "absolute_change_ej_per_year": 0.6, "percent_change": 2.0},
        {"technology": "Concentrated Solar Power", "absolute_change_ej_per_year": 7.5, "percent_change": 2.2},
        {"technology": "Onshore Wind", "absolute_change_ej_per_year": -22.1, "percent_change": -4.1},
        {"technology": "Offshore Wind", "absolute_change_ej_per_year": -8.6, "percent_change": -2.1},
        {"technology": "Hydropower", "absolute_change_ej_per_year": 2.2, "percent_change": 6.1},
        {"technology": "First-generation Bioenergy", "absolute_change_ej_per_year": 5.8, "percent_change": 32.4},
        {"technology": "Second-generation Bioenergy", "absolute_change_ej_per_year": 45.9, "percent_change": 38.1},
    ]
)

summary_df
```

```python
summary_stats = summary_df[["absolute_change_ej_per_year", "percent_change"]].describe().round(2)
summary_stats
```

```python
sns.set_theme(style="whitegrid")
ordered = summary_df.sort_values("percent_change", ascending=False)

# Build a per-technology colour palette (blue = positive, orange = negative)
palette = {
    row["technology"]: "#2c7fb8" if row["percent_change"] >= 0 else "#d95f0e"
    for _, row in ordered.iterrows()
}

plt.figure(figsize=(11, 6))
ax = sns.barplot(
    data=ordered,
    x="percent_change",
    y="technology",
    hue="technology",
    dodge=False,
    legend=False,
    palette=palette,
)
ax.axvline(0, color="black", linewidth=1)
ax.set_title("Reported Global Change in Renewable Energy Potential")
ax.set_xlabel("Percent change (%)")
ax.set_ylabel("Technology")

for index, value in enumerate(ordered["percent_change"]):
    x_position = value + 0.8 if value >= 0 else value - 2.4
    ax.text(x_position, index, f"{value:.1f}%", va="center", fontsize=9)

plt.tight_layout()
plt.show()
```

## Interpretation

The recreated summary figure highlights three takeaways that match the paper's broader discussion:

- **Solar power is comparatively stable** under projected climate change, with only small average changes.
- **Wind and hydropower are more uncertain and regionally variable**, so local planning matters more than global means.
- **Bioenergy shows the strongest positive change in some scenarios**, but this depends strongly on assumptions such as CO2 fertilization.

This makes the paper valuable not only as a literature review, but also as a planning framework for climate-resilient renewable energy systems.
